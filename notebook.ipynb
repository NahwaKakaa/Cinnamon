````python
import os
import shutil
import pandas as pd
from sklearn.model_selection import train_test_split

def get_base_name(fname):
    """
    Ekstrak group_id (nama dasar foto sebelum augmentasi) untuk mencegah data leakage.
    Roboflow menghasilkan varian augmentasi dengan format: <nama_asli>.rf.<hash>.<ekstensi>
    """
    if '.rf.' in fname:
        return fname.split('.rf.')[0]
    return os.path.splitext(fname)[0]

# Validasi fungsi get_base_name (group_id)
assert get_base_name("IMG_E4793_jpg.rf.66b3a30ca40300d98b0f71cd15f6068b.jpg") == "IMG_E4793_jpg", "Validasi get_base_name gagal!"
assert get_base_name("sample_photo.png") == "sample_photo", "Validasi fallback get_base_name gagal!"
print("✅ Fungsi get_base_name (group_id) tervalidasi dengan benar.")

orig_train_dir = '/kaggle/working/Final_Cinnamon-1/train'
new_base = '/kaggle/working/dataset_split_clean'

# Hapus hasil split lama agar tidak ada file sisa
if os.path.exists(new_base):
    shutil.rmtree(new_base)

# ============================================================
# 1. Kumpulkan SEMUA gambar dan kelompokkan berdasarkan base name
# ============================================================
groups = {}

for cls in sorted(os.listdir(orig_train_dir)):
    cls_path = os.path.join(orig_train_dir, cls)

    if not os.path.isdir(cls_path):
        continue

    for fname in os.listdir(cls_path):
        if not fname.lower().endswith(('.jpg', '.jpeg', '.png', '.bmp')):
            continue

        fpath = os.path.join(cls_path, fname)
        base_name = get_base_name(fname)

        groups.setdefault(base_name, []).append({
            'path': fpath,
            'class': cls,
            'fname': fname
        })

n_raw_images = sum(len(items) for items in groups.values())
print(f"Total gambar mentah : {n_raw_images}")
print(f"Total group unik    : {len(groups)}")

# ============================================================
# 2. Deteksi group yang memiliki konflik kelas
# ============================================================
conflicting_groups = []

for base_name, items in groups.items():
    classes = set(item['class'] for item in items)
    if len(classes) > 1:
        conflicting_groups.append({
            'base_name': base_name,
            'classes': sorted(classes),
            'files': [item['fname'] for item in items]
        })

print(f"Group dengan konflik kelas: {len(conflicting_groups)}")

# ============================================================
# 3. Keluarkan group konflik dari dataset split secara konsisten
# ============================================================
conflicting_base_names = {
    item['base_name']
    for item in conflicting_groups
}

# Alur gambar dikeluarkan vs dipakai
n_removed_images = sum(len(groups[b]) for b in conflicting_base_names)
n_kept_images = n_raw_images - n_removed_images
print(f"Group konflik dikeluarkan : {len(conflicting_base_names)} group ({n_removed_images} gambar)")
print(f"Gambar yang dipakai       : {n_kept_images}")

# Simpan metadata audit 184 file konflik ke CSV
conflict_records = []
for item in conflicting_groups:
    for f in item['files']:
        conflict_records.append({
            'group_id': item['base_name'],
            'conflicting_classes': ' | '.join(item['classes']),
            'filename': f
        })
df_conflicts = pd.DataFrame(conflict_records)
os.makedirs('/kaggle/working/cinnamon-output', exist_ok=True)
df_conflicts.to_csv('/kaggle/working/label_conflicts_184.csv', index=False)
df_conflicts.to_csv('/kaggle/working/cinnamon-output/label_conflicts_184.csv', index=False)

# Validasi ketat bahwa tepat 184 gambar konflik berhasil dikecualikan secara konsisten
assert n_removed_images == 184, f"Ekspektasi 184 gambar konflik dikecualikan, namun terhitung: {n_removed_images}"
print(f"✅ Konsistensi terverifikasi: Tepat {n_removed_images} gambar ({len(conflicting_groups)} grup) konflik berhasil dikecualikan secara konsisten.")
print("📄 Metadata audit konflik tersimpan di /kaggle/working/label_conflicts_184.csv")

print("\n" + "=" * 60)
print("PENANGANAN GROUP DENGAN KONFLIK LABEL")
print("=" * 60)

for base_name in sorted(conflicting_base_names):
    classes = next(
        item['classes']
        for item in conflicting_groups
        if item['base_name'] == base_name
    )
    print(f"  - {base_name}: {classes}")

print("\n⚠️ Group konflik TIDAK dihapus dari dataset asli.")
print("⚠️ Group konflik dikecualikan secara konsisten dari dataset_split_clean.")

# ============================================================
# 4. Buat daftar group berdasarkan kelas (Group konflik dilewati)
# ============================================================
class_groups = {}

for base_name, items in groups.items():
    if base_name in conflicting_base_names:
        continue
    cls = items[0]['class']
    class_groups.setdefault(cls, []).append(base_name)

print("\n===== JUMLAH GROUP SETELAH KONFLIK DIKELUARKAN =====")
for cls in sorted(class_groups.keys()):
    print(f"{cls}: {len(class_groups[cls])} groups")

# ============================================================
# 5. Split setiap kelas berdasarkan GROUP
#    80% train - 10% validation - 10% test
# ============================================================
split_map = {
    'train': {},
    'val': {},
    'test': {}
}

for cls in sorted(class_groups.keys()):
    base_names = class_groups[cls]

    train_bases, temp_bases = train_test_split(
        base_names,
        test_size=0.20,
        random_state=42
    )

    val_bases, test_bases = train_test_split(
        temp_bases,
        test_size=0.50,
        random_state=42
    )

    split_map['train'][cls] = train_bases
    split_map['val'][cls] = val_bases
    split_map['test'][cls] = test_bases

    print(
        f"{cls}: "
        f"train={len(train_bases)}, "
        f"val={len(val_bases)}, "
        f"test={len(test_bases)} groups"
    )

# ============================================================
# 6. Copy file ke folder train / val / test
# ============================================================
for split in ['train', 'val', 'test']:
    for cls, base_names in split_map[split].items():
        out_dir = os.path.join(new_base, split, cls)
        os.makedirs(out_dir, exist_ok=True)

        for base_name in base_names:
            for item in groups[base_name]:
                src = item['path']
                dst = os.path.join(out_dir, item['fname'])
                shutil.copy2(src, dst)

# ============================================================
# 7. Verifikasi jumlah gambar
# ============================================================
print("\n" + "=" * 60)
print("JUMLAH GAMBAR HASIL SPLIT")
print("=" * 60)

for split in ['train', 'val', 'test']:
    print(f"\n{split.upper()}")
    split_dir = os.path.join(new_base, split)
    total = 0

    for cls in sorted(os.listdir(split_dir)):
        cls_path = os.path.join(split_dir, cls)
        if not os.path.isdir(cls_path):
            continue

        count = len([
            f for f in os.listdir(cls_path)
            if os.path.isfile(os.path.join(cls_path, f))
        ])
        total += count
        print(f"  {cls}: {count}")
    print(f"  TOTAL: {total}")

# ============================================================
# 8. Verifikasi tidak ada group overlap / leakage
# ============================================================
train_groups = set()
val_groups = set()
test_groups = set()

for split, target_set in [
    ('train', train_groups),
    ('val', val_groups),
    ('test', test_groups)
]:
    split_dir = os.path.join(new_base, split)
    for cls in os.listdir(split_dir):
        cls_path = os.path.join(split_dir, cls)
        if not os.path.isdir(cls_path):
            continue
        for fname in os.listdir(cls_path):
            target_set.add(get_base_name(fname))

print("\n" + "=" * 60)
print("VERIFIKASI GROUP LEAKAGE")
print("=" * 60)

print("Train groups:", len(train_groups))
print("Val groups  :", len(val_groups))
print("Test groups :", len(test_groups))

print("Train ∩ Val :", len(train_groups & val_groups))
print("Train ∩ Test:", len(train_groups & test_groups))
print("Val ∩ Test  :", len(val_groups & test_groups))

# Validasi assert ketat
assert train_groups.isdisjoint(val_groups), "❌ GROUP LEAKAGE: Train dan Val saling berbagi group!"
assert train_groups.isdisjoint(test_groups), "❌ GROUP LEAKAGE: Train dan Test saling berbagi group!"
assert val_groups.isdisjoint(test_groups), "❌ GROUP LEAKAGE: Val dan Test saling berbagi group!"

# Pastikan tidak ada satupun group konflik yang masuk ke split
for s_name, s_groups in [('Train', train_groups), ('Val', val_groups), ('Test', test_groups)]:
    conflicts_in_subset = s_groups & conflicting_base_names
    assert len(conflicts_in_subset) == 0, f"❌ LEAKAGE: {len(conflicts_in_subset)} group konflik masuk ke {s_name}!"

# Verifikasi seluruh gambar dari tiap group berada utuh di subset yang sama
for split, b_set in [('train', train_groups), ('val', val_groups), ('test', test_groups)]:
    for b in b_set:
        for item in groups[b]:
            expected = os.path.join(new_base, split, item['class'], item['fname'])
            assert os.path.exists(expected), f"File {item['fname']} tidak ditemukan pada subset {split}!"

print("\n✅ Validasi group_id & subset tuntas: Tidak ada group yang overlap.")
print("✅ Tepat 184 gambar dari 19 group konflik telah dikecualikan secara konsisten.")
print("✅ Dataset berhasil dibuat dengan validasi group-based splitting.")

# ============================================================
# STATISTIK GROUP DAN GAMBAR PER SUBSET
# ============================================================
rows = []
for split in ['train', 'val', 'test']:
    for cls in sorted(split_map[split]):
        bases = split_map[split][cls]
        rows.append({
            'split': split, 'class': cls,
            'groups': len(bases),
            'images': sum(len(groups[b]) for b in bases)
        })
df_group_stats = pd.DataFrame(rows)
print("\n===== STATISTIK JUMLAH GROUP & GAMBAR PER SUBSET =====")
print(df_group_stats.pivot(index='class', columns='split', values=['groups', 'images']))
df_group_stats.to_csv('/kaggle/working/group_split_statistics.csv', index=False)
df_group_stats.to_csv('/kaggle/working/cinnamon-output/group_split_statistics.csv', index=False)
assert n_kept_images == df_group_stats['images'].sum()
````

````python
import cv2
import numpy as np
import os
import random
import pandas as pd
import matplotlib.pyplot as plt

def compute_quality_scores(image_path):
    img = cv2.imread(image_path)
    if img is None:
        return None, None
    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
    blur_score = cv2.Laplacian(gray, cv2.CV_64F).var()
    denoised = cv2.medianBlur(gray, 5)
    noise_score = np.std(gray.astype(np.float32) - denoised.astype(np.float32))
    return blur_score, noise_score

def scan_dataset_quality(root_dir, blur_threshold=15, noise_threshold=25):
    flagged = []
    for cls in sorted(os.listdir(root_dir)):
        cls_path = os.path.join(root_dir, cls)
        if not os.path.isdir(cls_path):
            continue
        for fname in os.listdir(cls_path):
            fpath = os.path.join(cls_path, fname)
            blur_score, noise_score = compute_quality_scores(fpath)
            if blur_score is None:
                continue
            if blur_score < blur_threshold or noise_score > noise_threshold:
                flagged.append({'path': fpath, 'class': cls,
                                 'blur_score': blur_score, 'noise_score': noise_score})
    return flagged

train_dir_check = '/kaggle/working/dataset_split_clean/train'
flagged_images = scan_dataset_quality(train_dir_check)

total = sum(len(os.listdir(os.path.join(train_dir_check, c))) for c in os.listdir(train_dir_check))
print(f"Total gambar training: {total}")
print(f"Gambar ditandai rusak berat (blur/noise ekstrem): {len(flagged_images)}")

if flagged_images:
    df_flagged = pd.DataFrame(flagged_images)
    print(df_flagged.groupby('class').size())

    sample_check = random.sample(flagged_images, min(12, len(flagged_images)))
    plt.figure(figsize=(15, 10))
    for i, item in enumerate(sample_check):
        img = cv2.cvtColor(cv2.imread(item['path']), cv2.COLOR_BGR2RGB)
        plt.subplot(3, 4, i + 1)
        plt.imshow(img)
        plt.title(f"{item['class']}\nblur={item['blur_score']:.1f} noise={item['noise_score']:.1f}", fontsize=8)
        plt.axis('off')
    plt.tight_layout()
    plt.show()
    print("\n👀 Cek dulu gambar di atas. Kalau memang rusak, jalankan cell berikutnya untuk menghapus.")
else:
    print("Tidak ada gambar yang ditandai rusak.")
````

````python
# ============================================================
# PERBANDINGAN 4 ARSITEKTUR
# ============================================================

import os
import json
import gc
import shutil
import numpy as np
import pandas as pd
import tensorflow as tf
import matplotlib.pyplot as plt

from tensorflow.keras.applications import (
    ResNet50,
    EfficientNetB0,
    MobileNetV2,
    DenseNet121
)

from tensorflow.keras.applications.resnet50 import (
    preprocess_input as resnet_preprocess
)

from tensorflow.keras.applications.efficientnet import (
    preprocess_input as efficientnet_preprocess
)

from tensorflow.keras.applications.mobilenet_v2 import (
    preprocess_input as mobilenet_preprocess
)

from tensorflow.keras.applications.densenet import (
    preprocess_input as densenet_preprocess
)

from tensorflow.keras.models import Model
from tensorflow.keras.layers import (
    Dense,
    GlobalAveragePooling2D,
    Dropout
)

from tensorflow.keras.optimizers import Adam

from tensorflow.keras.callbacks import (
    EarlyStopping,
    ReduceLROnPlateau,
    ModelCheckpoint,
    BackupAndRestore
)

from tensorflow.keras import regularizers
from tensorflow.keras.preprocessing.image import ImageDataGenerator

from sklearn.utils.class_weight import compute_class_weight


# ============================================================
# PATH
# ============================================================

train_dir = '/kaggle/working/dataset_split_clean/train'
val_dir = '/kaggle/working/dataset_split_clean/val'
test_dir = '/kaggle/working/dataset_split_clean/test'

RESULTS_FILE = '/kaggle/working/model_comparison_results.json'
HISTORY_FILE = '/kaggle/working/model_comparison_histories.json'

BACKUP_DIR = '/kaggle/working/model_backups'
PLOT_DIR = '/kaggle/working/model_comparison_plots'

os.makedirs(BACKUP_DIR, exist_ok=True)
os.makedirs(PLOT_DIR, exist_ok=True)


# ============================================================
# MODEL CONFIGURATION
# ============================================================

MODEL_CONFIGS = {

    'ResNet50': {
        'base_class': ResNet50,
        'preprocess': resnet_preprocess
    },

    'EfficientNetB0': {
        'base_class': EfficientNetB0,
        'preprocess': efficientnet_preprocess
    },

    'MobileNetV2': {
        'base_class': MobileNetV2,
        'preprocess': mobilenet_preprocess
    },

    'DenseNet121': {
        'base_class': DenseNet121,
        'preprocess': densenet_preprocess
    }
}


# ============================================================
# PARAMETER
# ============================================================

IMG_SIZE = (224, 224)
BATCH_SIZE = 32

PHASE1_EPOCHS = 10
PHASE2_EPOCHS = 20

UNFREEZE_N = 20

LEARNING_RATE_PHASE1 = 0.001
LEARNING_RATE_PHASE2 = 1e-5

NUM_CLASSES = 6


# ============================================================
# BUILD MODEL
# ============================================================

def build_comparison_model(base_class):

    base = base_class(
        weights='imagenet',
        include_top=False,
        input_shape=(224, 224, 3)
    )

    # Phase 1: backbone dibekukan
    base.trainable = False

    x = base.output

    x = GlobalAveragePooling2D()(x)

    x = Dense(
        512,
        activation='relu',
        kernel_regularizer=regularizers.l2(0.005)
    )(x)

    x = Dropout(0.6)(x)

    output = Dense(
        NUM_CLASSES,
        activation='softmax'
    )(x)

    model = Model(
        inputs=base.input,
        outputs=output
    )

    model.compile(
        optimizer=Adam(
            learning_rate=LEARNING_RATE_PHASE1
        ),
        loss='categorical_crossentropy',
        metrics=['accuracy']
    )

    return model, base


# ============================================================
# GENERATOR
# ============================================================

def make_comparison_generators(preprocess_fn):

    train_datagen = ImageDataGenerator(
        preprocessing_function=preprocess_fn,
        rotation_range=20,
        zoom_range=0.2,
        horizontal_flip=True,
        fill_mode='nearest'
    )

    eval_datagen = ImageDataGenerator(
        preprocessing_function=preprocess_fn
    )

    train_gen = train_datagen.flow_from_directory(
        train_dir,
        target_size=IMG_SIZE,
        batch_size=BATCH_SIZE,
        class_mode='categorical',
        shuffle=True,
        seed=42
    )

    val_gen = eval_datagen.flow_from_directory(
        val_dir,
        target_size=IMG_SIZE,
        batch_size=BATCH_SIZE,
        class_mode='categorical',
        shuffle=False
    )

    return train_gen, val_gen


# ============================================================
# RESET HASIL COMPARISON LAMA
# ============================================================
# Hasil comparison lama dihapus agar 4 model benar-benar
# dievaluasi berdasarkan training baru.
#
# BackupAndRestore TIDAK dihapus, sehingga jika runtime
# terputus, training dapat mencoba melanjutkan dari backup.

if os.path.exists(RESULTS_FILE):
    os.remove(RESULTS_FILE)

if os.path.exists(HISTORY_FILE):
    os.remove(HISTORY_FILE)


# ============================================================
# HASIL
# ============================================================

comparison_results = {}
comparison_histories = {}


# ============================================================
# TRAINING 4 ARSITEKTUR
# ============================================================

for model_name, config in MODEL_CONFIGS.items():

    print("\n" + "=" * 70)
    print(f"TRAINING MODEL: {model_name}")
    print("=" * 70)


    # --------------------------------------------------------
    # CHECKPOINT
    # --------------------------------------------------------

    checkpoint_path = (
        f'/kaggle/working/'
        f'best_{model_name}_comparison.keras'
    )


    # --------------------------------------------------------
    # GENERATOR
    # --------------------------------------------------------

    train_gen, val_gen = make_comparison_generators(
        config['preprocess']
    )


    # --------------------------------------------------------
    # CLASS WEIGHT
    # --------------------------------------------------------

    class_weights = compute_class_weight(
        class_weight='balanced',
        classes=np.unique(train_gen.classes),
        y=train_gen.classes
    )

    class_weight_dict = dict(
        enumerate(class_weights)
    )


    # --------------------------------------------------------
    # BUILD MODEL
    # --------------------------------------------------------

    model, base = build_comparison_model(
        config['base_class']
    )


    # ========================================================
    # PHASE 1
    # ========================================================

    print(f"\n{model_name} - PHASE 1")

    phase1_backup = os.path.join(
        BACKUP_DIR,
        f'{model_name}_phase1'
    )

    history_phase1 = model.fit(
        train_gen,
        epochs=PHASE1_EPOCHS,
        validation_data=val_gen,
        class_weight=class_weight_dict,

        callbacks=[
            EarlyStopping(
                monitor='val_loss',
                patience=3,
                restore_best_weights=True,
                verbose=1
            ),

            BackupAndRestore(
                backup_dir=phase1_backup,
                delete_checkpoint=False
            )
        ],

        verbose=1
    )


    # ========================================================
    # PHASE 2 - FINE TUNING
    # ========================================================

    print(f"\n{model_name} - PHASE 2 FINE-TUNING")

    base.trainable = True


    # Bekukan semua layer kecuali N layer terakhir
    for layer in base.layers[:-UNFREEZE_N]:
        layer.trainable = False


    # Aktifkan N layer terakhir
    for layer in base.layers[-UNFREEZE_N:]:
        layer.trainable = True


    # --------------------------------------------------------
    # BATCH NORMALIZATION TETAP BEKU
    # --------------------------------------------------------

    for layer in base.layers[-UNFREEZE_N:]:

        if isinstance(
            layer,
            tf.keras.layers.BatchNormalization
        ):
            layer.trainable = False


    # --------------------------------------------------------
    # COMPILE ULANG
    # --------------------------------------------------------

    model.compile(
        optimizer=Adam(
            learning_rate=LEARNING_RATE_PHASE2
        ),
        loss='categorical_crossentropy',
        metrics=['accuracy']
    )


    phase2_backup = os.path.join(
        BACKUP_DIR,
        f'{model_name}_phase2'
    )


    # ========================================================
    # PHASE 2 TRAINING
    # ========================================================

    history_phase2 = model.fit(
        train_gen,
        epochs=PHASE2_EPOCHS,
        validation_data=val_gen,
        class_weight=class_weight_dict,

        callbacks=[
            EarlyStopping(
                monitor='val_loss',
                patience=4,
                restore_best_weights=True,
                verbose=1
            ),

            ReduceLROnPlateau(
                monitor='val_loss',
                factor=0.5,
                patience=2,
                min_lr=1e-7,
                verbose=1
            ),

            ModelCheckpoint(
                checkpoint_path,
                monitor='val_accuracy',
                save_best_only=True,
                verbose=1
            ),

            BackupAndRestore(
                backup_dir=phase2_backup,
                delete_checkpoint=False
            )
        ],

        verbose=1
    )


    # ========================================================
    # GABUNGKAN HISTORY PHASE 1 + PHASE 2
    # ========================================================

    history = {}

    for key in history_phase1.history.keys():

        history[key] = [
            float(x)
            for x in history_phase1.history[key]
        ]


    for key in history_phase2.history.keys():

        if key not in history:
            history[key] = []

        history[key].extend([
            float(x)
            for x in history_phase2.history[key]
        ])


    comparison_histories[model_name] = history


    # ========================================================
    # SIMPAN HISTORY
    # ========================================================

    with open(
        HISTORY_FILE,
        'w'
    ) as f:

        json.dump(
            comparison_histories,
            f,
            indent=2
        )


    # ========================================================
    # LOAD BEST MODEL
    # ========================================================

    if os.path.exists(checkpoint_path):

        best_model = tf.keras.models.load_model(
            checkpoint_path
        )

    else:

        best_model = model


    # ========================================================
    # VALIDATION EVALUATION
    # ========================================================

    val_loss, val_acc = best_model.evaluate(
        val_gen,
        verbose=0
    )


    print(
        f"\n{model_name} - "
        f"Validation Accuracy: "
        f"{val_acc * 100:.2f}%"
    )


    # ========================================================
    # SIMPAN HASIL COMPARISON
    # ========================================================

    comparison_results[model_name] = {

        'validation_accuracy': float(
            val_acc
        ),

        'validation_loss': float(
            val_loss
        ),

        'checkpoint': checkpoint_path,

        'total_params': int(
            best_model.count_params()
        )
    }


    with open(
        RESULTS_FILE,
        'w'
    ) as f:

        json.dump(
            comparison_results,
            f,
            indent=2
        )


    print(
        f"✅ Hasil {model_name} tersimpan."
    )


    # ========================================================
    # CLEAN MEMORY
    # ========================================================

    del best_model
    del model
    del base
    del train_gen
    del val_gen

    tf.keras.backend.clear_session()
    gc.collect()


# ============================================================
# HASIL AKHIR PERBANDINGAN
# ============================================================

df_validation = (
    pd.DataFrame(comparison_results)
    .T
    .sort_values(
        'validation_accuracy',
        ascending=False
    )
)


print("\n" + "=" * 70)
print("HASIL PERBANDINGAN 4 ARSITEKTUR")
print("=" * 70)

print(df_validation)


# ============================================================
# MODEL TERBAIK
# ============================================================

best_model_name = df_validation.index[0]

best_model_path = comparison_results[
    best_model_name
]['checkpoint']

best_validation_accuracy = comparison_results[
    best_model_name
]['validation_accuracy']


print("\n" + "=" * 70)

print(
    f"MODEL TERBAIK : {best_model_name}"
)

print(
    f"VALIDATION ACCURACY : "
    f"{best_validation_accuracy * 100:.2f}%"
)

print(
    f"CHECKPOINT : {best_model_path}"
)

print("=" * 70)


# ============================================================
# GRAFIK TRAINING
# ============================================================

for model_name, history in comparison_histories.items():

    epochs = range(
        1,
        len(history['accuracy']) + 1
    )


    # ========================================================
    # ACCURACY
    # ========================================================

    plt.figure(figsize=(10, 6))

    plt.plot(
        epochs,
        history['accuracy'],
        label='Training Accuracy'
    )

    plt.plot(
        epochs,
        history['val_accuracy'],
        label='Validation Accuracy'
    )

    plt.xlabel('Epoch')
    plt.ylabel('Accuracy')

    plt.title(
        f'{model_name} - Training and Validation Accuracy'
    )

    plt.legend()
    plt.grid(True)

    plt.tight_layout()

    accuracy_path = os.path.join(
        PLOT_DIR,
        f'{model_name}_accuracy.png'
    )

    plt.savefig(
        accuracy_path,
        dpi=300,
        bbox_inches='tight'
    )

    plt.show()

    plt.close()


    # ========================================================
    # LOSS
    # ========================================================

    plt.figure(figsize=(10, 6))

    plt.plot(
        epochs,
        history['loss'],
        label='Training Loss'
    )

    plt.plot(
        epochs,
        history['val_loss'],
        label='Validation Loss'
    )

    plt.xlabel('Epoch')
    plt.ylabel('Loss')

    plt.title(
        f'{model_name} - Training and Validation Loss'
    )

    plt.legend()
    plt.grid(True)

    plt.tight_layout()

    loss_path = os.path.join(
        PLOT_DIR,
        f'{model_name}_loss.png'
    )

    plt.savefig(
        loss_path,
        dpi=300,
        bbox_inches='tight'
    )

    plt.show()

    plt.close()


# ============================================================
# GRAFIK PERBANDINGAN VALIDATION ACCURACY
# ============================================================

plt.figure(figsize=(10, 6))

plt.bar(
    df_validation.index,
    df_validation['validation_accuracy'] * 100
)

plt.xlabel('Architecture')
plt.ylabel('Validation Accuracy (%)')

plt.title(
    'Validation Accuracy Comparison of Four Architectures'
)

plt.xticks(rotation=15)

plt.tight_layout()

comparison_plot_path = os.path.join(
    PLOT_DIR,
    'architecture_validation_accuracy_comparison.png'
)

plt.savefig(
    comparison_plot_path,
    dpi=300,
    bbox_inches='tight'
)

plt.show()

plt.close()


# ============================================================
# INFORMASI OUTPUT
# ============================================================

print("\nFile hasil training:")

print(
    f"Results : {RESULTS_FILE}"
)

print(
    f"History : {HISTORY_FILE}"
)

print(
    f"Plots   : {PLOT_DIR}"
)

print(
    f"Best Model : {best_model_path}"
)
````

````python
# ============================================================
# FINAL EXTERNAL TEST
# Model terbaik dipilih otomatis berdasarkan validation accuracy
# ============================================================

import os
import json
import numpy as np
import pandas as pd
import tensorflow as tf
import matplotlib.pyplot as plt
import seaborn as sns

from PIL import Image

from sklearn.metrics import (
    classification_report,
    confusion_matrix,
    accuracy_score
)


# ============================================================
# PATH & DATASET DISCOVERY / AUTO-DOWNLOAD
# ============================================================

CSV_PATH = None

# 1. Cari di /kaggle/input (diutamakan folder 'test')
if os.path.exists('/kaggle/input'):
    for root, dirs, files in os.walk('/kaggle/input'):
        if '_classes.csv' in files and os.path.basename(root).lower() == 'test':
            CSV_PATH = os.path.join(root, '_classes.csv')
            break
    if CSV_PATH is None:
        for root, dirs, files in os.walk('/kaggle/input'):
            if '_classes.csv' in files:
                CSV_PATH = os.path.join(root, '_classes.csv')
                break

# 2. Cari di /kaggle/working jika belum ditemukan di /kaggle/input
if CSV_PATH is None and os.path.exists('/kaggle/working'):
    for root, dirs, files in os.walk('/kaggle/working'):
        if '_classes.csv' in files and os.path.basename(root).lower() == 'test':
            CSV_PATH = os.path.join(root, '_classes.csv')
            break
    if CSV_PATH is None:
        for root, dirs, files in os.walk('/kaggle/working'):
            if '_classes.csv' in files:
                CSV_PATH = os.path.join(root, '_classes.csv')
                break

# 3. Jika belum ditemukan di mana pun, unduh otomatis via Roboflow API / Kaggle Secrets
if CSV_PATH is None:
    print("Dataset external test tidak ditemukan di /kaggle/input maupun /kaggle/working.")
    print("Mempersiapkan pengunduhan dataset external...")

    # Mengambil API key secara aman dari Kaggle Secrets
    api_key_ext = None
    try:
        from kaggle_secrets import UserSecretsClient
        user_secrets = UserSecretsClient()
        api_key_ext = user_secrets.get_secret("ROBOFLOW_API_KEY")
        print("✅ Berhasil memuat ROBOFLOW_API_KEY dari Kaggle Secrets.")
    except Exception as e_sec:
        api_key_ext = os.environ.get("ROBOFLOW_API_KEY", "")
        if not api_key_ext:
            print(f"⚠️ Kaggle Secrets tidak terdeteksi ({e_sec}).")
            print("💡 Pastikan 'ROBOFLOW_API_KEY' telah ditambahkan di Add-ons -> Secrets.")

    # Metode A: Unduh melalui Roboflow SDK dengan API Key dari Secrets
    if api_key_ext:
        try:
            from roboflow import Roboflow
            rf = Roboflow(api_key=api_key_ext)
            ext_proj = rf.workspace("kalhara-vithanage").project("cinnamon-multiple-diseases")
            ext_ds = ext_proj.version(1).download("multiclass")
            print("✅ Berhasil mengunduh dataset external melalui Roboflow API.")
        except Exception as e_rf:
            print(f"Download via Roboflow API dilewati ({e_rf}).")

    # Metode B: Fallback download terotentikasi dinamis (tanpa hardcoded secret)
    ext_dir_working = '/kaggle/working/cinnamon-multiple-diseases-1'
    has_csv_working = os.path.exists(ext_dir_working) and any(
        '_classes.csv' in f for _, _, f in os.walk(ext_dir_working)
    )
    if not has_csv_working and api_key_ext:
        try:
            import urllib.request, zipfile, io, ssl
            print("Mencoba unduh dataset external terotentikasi via URL endpoint...")
            zip_url = f"https://app.roboflow.com/ds/mEOKFzSdwv?key={api_key_ext}"
            os.makedirs(ext_dir_working, exist_ok=True)
            ctx = ssl._create_unverified_context()
            req = urllib.request.Request(zip_url, headers={"User-Agent": "Mozilla/5.0"})
            with urllib.request.urlopen(req, context=ctx, timeout=60) as resp:
                z = zipfile.ZipFile(io.BytesIO(resp.read()))
                z.extractall(ext_dir_working)
            print("✅ Ekstraksi dataset external berhasil!")
        except Exception as e_zip:
            print(f"Download fallback endpoint gagal: {e_zip}")

    # Cari ulang di /kaggle/working
    for root, dirs, files in os.walk('/kaggle/working'):
        if '_classes.csv' in files and os.path.basename(root).lower() == 'test':
            CSV_PATH = os.path.join(root, '_classes.csv')
            break
    if CSV_PATH is None:
        for root, dirs, files in os.walk('/kaggle/working'):
            if '_classes.csv' in files:
                CSV_PATH = os.path.join(root, '_classes.csv')
                break

if CSV_PATH is None or not os.path.exists(CSV_PATH):
    raise FileNotFoundError(
        "Gagal menemukan atau mengunduh dataset external test (_classes.csv)."
    )

EXTERNAL_TEST_DIR = os.path.dirname(CSV_PATH)
print("✅ CSV External Test ditemukan:", CSV_PATH)
print("✅ Direktori External Test     :", EXTERNAL_TEST_DIR)

RESULTS_FILE = (
    "/kaggle/working/model_comparison_results.json"
)

OUTPUT_DIR = (
    "/kaggle/working/visualisasi_hasil"
)

os.makedirs(
    OUTPUT_DIR,
    exist_ok=True
)

# ============================================================
# LOAD HASIL COMPARISON
# ============================================================

if not os.path.exists(RESULTS_FILE):

    raise FileNotFoundError(
        f"File hasil comparison tidak ditemukan:\n"
        f"{RESULTS_FILE}"
    )

with open(
    RESULTS_FILE,
    "r"
) as f:

    comparison_results = json.load(f)


# ============================================================
# PILIH MODEL TERBAIK OTOMATIS
# ============================================================

best_model_name = max(
    comparison_results,
    key=lambda x:
    comparison_results[x]["validation_accuracy"]
)

checkpoint_path = comparison_results[
    best_model_name
]["checkpoint"]

best_validation_accuracy = (
    comparison_results[
        best_model_name
    ]["validation_accuracy"]
)


print("=" * 70)
print("MODEL TERBAIK UNTUK EXTERNAL TEST")
print("=" * 70)

print(
    f"Best Model            : {best_model_name}"
)

print(
    f"Validation Accuracy   : "
    f"{best_validation_accuracy * 100:.2f}%"
)

print(
    f"Checkpoint            : {checkpoint_path}"
)


if not os.path.exists(checkpoint_path):

    raise FileNotFoundError(
        f"Checkpoint tidak ditemukan:\n"
        f"{checkpoint_path}"
    )


# ============================================================
# CLASS MAPPING
# ============================================================

external_to_model = {

    "healthy_cinnamon":
        "healthy",

    "leaf_spot_disease":
        "leafspot",

    "rough_bark":
        "roughbark",

    "stripe_canker":
        "stripecanker"
}


model_classes = [

    "healthy",
    "leafspot",
    "noncinnamon_bark",
    "noncinnamon_leaves",
    "roughbark",
    "stripecanker"
]


model_class_to_idx = {

    name: idx

    for idx, name in enumerate(
        model_classes
    )
}


external_class_names = [

    "healthy",
    "leafspot",
    "roughbark",
    "stripecanker"
]


external_model_indices = [

    model_class_to_idx["healthy"],
    model_class_to_idx["leafspot"],
    model_class_to_idx["roughbark"],
    model_class_to_idx["stripecanker"]
]


label_cols = [

    "healthy_cinnamon",
    "leaf_spot_disease",
    "rough_bark",
    "stripe_canker"
]


# ============================================================
# LOAD CSV
# ============================================================

if not os.path.exists(CSV_PATH):

    raise FileNotFoundError(
        f"_classes.csv tidak ditemukan:\n"
        f"{CSV_PATH}"
    )

df_ext = pd.read_csv(
    CSV_PATH
)

df_ext.columns = (
    df_ext.columns
    .astype(str)
    .str.strip()
)


print(
    f"\nJumlah data dalam CSV : {len(df_ext)}"
)

print(
    "Kolom CSV:",
    list(df_ext.columns)
)


# ============================================================
# VALIDASI KOLOM
# ============================================================

required_columns = [
    "filename"
] + label_cols

missing_columns = [

    col

    for col in required_columns

    if col not in df_ext.columns
]

if missing_columns:

    raise ValueError(
        "Kolom berikut tidak ditemukan:\n"
        + str(missing_columns)
    )


# ============================================================
# GROUND TRUTH
# ============================================================

y_true = []

image_paths = []

used_filenames = []

skipped_rows = 0


for _, row in df_ext.iterrows():

    filename = str(
        row["filename"]
    ).strip()


    # --------------------------------------------------------
    # Cari label aktif
    # --------------------------------------------------------

    active_labels = []

    for col in label_cols:

        try:

            value = float(
                row[col]
            )

            if value == 1:

                active_labels.append(
                    col
                )

        except:

            value = (
                str(row[col])
                .strip()
                .lower()
            )

            if value in [
                "1",
                "1.0",
                "true"
            ]:

                active_labels.append(
                    col
                )


    # Tepat satu label
    if len(active_labels) != 1:

        skipped_rows += 1
        continue


    external_label = (
        active_labels[0]
    )

    model_label = (
        external_to_model[
            external_label
        ]
    )


    # --------------------------------------------------------
    # Cari gambar
    # --------------------------------------------------------

    candidates = [

        os.path.join(
            EXTERNAL_TEST_DIR,
            filename
        ),

        os.path.join(
            EXTERNAL_TEST_DIR,
            os.path.basename(filename)
        )
    ]


    image_path = None


    for candidate in candidates:

        if os.path.exists(candidate):

            image_path = candidate
            break


    # Jika belum ditemukan, cari recursive
    if image_path is None:

        basename = os.path.basename(
            filename
        )

        for root, dirs, files in os.walk(
            EXTERNAL_TEST_DIR
        ):

            if basename in files:

                image_path = os.path.join(
                    root,
                    basename
                )

                break


    if image_path is None:

        skipped_rows += 1
        continue


    # --------------------------------------------------------
    # Simpan
    # --------------------------------------------------------

    image_paths.append(
        image_path
    )

    used_filenames.append(
        filename
    )

    y_true.append(
        model_class_to_idx[
            model_label
        ]
    )


print(
    f"\nGambar siap dievaluasi : "
    f"{len(image_paths)}"
)

print(
    f"Baris dilewati         : "
    f"{skipped_rows}"
)


if len(image_paths) == 0:

    raise ValueError(
        "Tidak ada gambar external yang dapat dievaluasi."
    )


# ============================================================
# LOAD MODEL
# ============================================================

print(
    f"\nMemuat model {best_model_name}..."
)

final_model = tf.keras.models.load_model(
    checkpoint_path
)

print(
    "Model berhasil dimuat."
)


# ============================================================
# PREPROCESSING SESUAI MODEL
# ============================================================

IMG_SIZE = (
    224,
    224
)


if best_model_name == "ResNet50":

    from tensorflow.keras.applications.resnet50 import (
        preprocess_input
    )

elif best_model_name == "MobileNetV2":

    from tensorflow.keras.applications.mobilenet_v2 import (
        preprocess_input
    )

elif best_model_name == "DenseNet121":

    from tensorflow.keras.applications.densenet import (
        preprocess_input
    )

elif best_model_name == "EfficientNetB0":

    from tensorflow.keras.applications.efficientnet import (
        preprocess_input
    )

else:

    raise ValueError(
        f"Model tidak dikenali: "
        f"{best_model_name}"
    )


# ============================================================
# LOAD GAMBAR
# ============================================================

X_external = []

valid_y_true = []

valid_paths = []


for image_path, true_label in zip(
    image_paths,
    y_true
):

    try:

        img = Image.open(
            image_path
        ).convert("RGB")

        img = img.resize(
            IMG_SIZE
        )

        img_array = np.asarray(
            img,
            dtype=np.float32
        )

        X_external.append(
            img_array
        )

        valid_y_true.append(
            true_label
        )

        valid_paths.append(
            image_path
        )

    except Exception as e:

        print(
            f"Gagal membaca: "
            f"{image_path}"
        )

        print(
            f"Error: {e}"
        )


X_external = np.asarray(
    X_external,
    dtype=np.float32
)

y_true = np.asarray(
    valid_y_true,
    dtype=np.int64
)

import collections
ext_counts = collections.Counter(model_classes[i] for i in y_true)
print("Jumlah gambar external per kelas:", dict(ext_counts))

image_paths = valid_paths


print(
    "\nShape data external:"
)

print(
    X_external.shape
)


# ============================================================
# PREPROCESS
# ============================================================

X_processed = preprocess_input(
    X_external
)


# ============================================================
# PREDICTION
# ============================================================

print(
    "\nMelakukan prediksi..."
)

y_prob = final_model.predict(
    X_processed,
    batch_size=32,
    verbose=1
)

y_pred = np.argmax(
    y_prob,
    axis=1
)

df_pred = pd.DataFrame({
    'file': [os.path.basename(p) for p in image_paths],
    'true': [model_classes[i] for i in y_true],
    'pred': [model_classes[i] for i in y_pred],
    'confidence': y_prob.max(axis=1)
})
df_pred['correct'] = df_pred['true'] == df_pred['pred']
df_pred.to_csv(os.path.join(OUTPUT_DIR, 'external_predictions.csv'), index=False)
print("Akurasi prediksi per kelas eksternal:")
print(df_pred.groupby('true')['correct'].agg(['sum', 'count']))


# ============================================================
# ACCURACY
# ============================================================

external_accuracy = accuracy_score(
    y_true,
    y_pred
)


print(
    "\n" + "=" * 70
)

print(
    f"EXTERNAL TEST - {best_model_name}"
)

print(
    "=" * 70
)

print(
    f"Jumlah gambar : "
    f"{len(y_true)}"
)

print(
    f"Accuracy      : "
    f"{external_accuracy:.4f}"
)

print(
    f"Accuracy (%)  : "
    f"{external_accuracy * 100:.2f}%"
)


# ============================================================
# METRIK EVALUASI LENGKAP: PRECISION, RECALL, F1, PER-CLASS CM
# ============================================================

def compute_per_class_metrics(y_true_arr, y_pred_arr, class_names_list, labels_list):
    cm_mat = confusion_matrix(y_true_arr, y_pred_arr, labels=labels_list)
    rep_dict = classification_report(
        y_true_arr, y_pred_arr, labels=labels_list, target_names=class_names_list,
        output_dict=True, zero_division=0
    )
    metric_rows = []
    for i, cname in enumerate(class_names_list):
        tp = cm_mat[i, i]
        fn = cm_mat[i, :].sum() - tp
        fp = cm_mat[:, i].sum() - tp
        tn = cm_mat.sum() - (tp + fn + fp)
        prec = rep_dict[cname]['precision']
        rec = rep_dict[cname]['recall']
        f1 = rep_dict[cname]['f1-score']
        supp = rep_dict[cname]['support']
        spec = tn / (tn + fp) if (tn + fp) > 0 else 0.0
        metric_rows.append({
            'Class': cname,
            'TP': int(tp), 'FP': int(fp), 'FN': int(fn), 'TN': int(tn),
            'Precision': round(float(prec), 4),
            'Recall (Sensitivity)': round(float(rec), 4),
            'Specificity': round(float(spec), 4),
            'F1-Score': round(float(f1), 4),
            'Support': int(supp)
        })
    return pd.DataFrame(metric_rows), cm_mat

print("\n" + "=" * 70)
print("METRIK EVALUASI EKSTERNAL PER KELAS")
print("=" * 70)

df_per_class_ext, _ = compute_per_class_metrics(
    y_true, y_pred, external_class_names, external_model_indices
)
print(df_per_class_ext.to_string(index=False))

# Simpan metrik per kelas ke CSV
df_per_class_ext.to_csv(os.path.join(OUTPUT_DIR, 'per_class_metrics_external.csv'), index=False)
df_per_class_ext.to_csv('/kaggle/working/per_class_metrics_external.csv', index=False)
print("✅ Metrik per kelas eksternal disimpan di per_class_metrics_external.csv")

print("\nClassification Report:")
print(
    classification_report(
        y_true,
        y_pred,
        labels=external_model_indices,
        target_names=external_class_names,
        digits=4,
        zero_division=0
    )
)


# ============================================================
# CONFUSION MATRIX
# ============================================================
# Menampilkan SEMUA 6 kelas output model.
# Dengan demikian jika external image diprediksi sebagai
# noncinnamon_bark atau noncinnamon_leaves, kesalahan tersebut
# tetap terlihat pada matrix.
# ============================================================

cm_external = confusion_matrix(
    y_true,
    y_pred,
    labels=np.arange(
        len(model_classes)
    )
)


plt.figure(
    figsize=(10, 7)
)

sns.heatmap(
    cm_external,
    annot=True,
    fmt="d",
    cmap="Blues",
    cbar=True,
    linewidths=1,
    linecolor="white",
    xticklabels=model_classes,
    yticklabels=model_classes,
    annot_kws={
        "fontsize": 12,
        "fontweight": "bold"
    }
)

plt.title(
    f"Confusion Matrix - External Test\n"
    f"{best_model_name}",
    fontsize=14,
    fontweight="bold",
    pad=15
)

plt.xlabel(
    "Predicted Class",
    fontsize=12
)

plt.ylabel(
    "True Class",
    fontsize=12
)

plt.xticks(
    rotation=35,
    ha="right",
    fontsize=10
)

plt.yticks(
    rotation=0,
    fontsize=10
)

plt.tight_layout()


external_cm_path = os.path.join(
    OUTPUT_DIR,
    "confusion_matrix_external_test.png"
)

plt.savefig(
    external_cm_path,
    dpi=300,
    bbox_inches="tight",
    facecolor="white"
)

plt.show()


# ============================================================
# SIMPAN HASIL EXTERNAL TEST
# ============================================================

external_results = {

    "model":
        best_model_name,

    "validation_accuracy":
        float(best_validation_accuracy),

    "external_accuracy":
        float(external_accuracy),

    "number_of_images":
        int(len(y_true)),

    "class_names":
        external_class_names,

    "model_classes":
        model_classes,

    "class_counts":
        dict(ext_counts),

    "per_class_metrics":
        df_per_class_ext.to_dict(orient="records"),

    "confusion_matrix":
        cm_external.tolist(),

    "checkpoint":
        checkpoint_path
}


external_results_path = os.path.join(
    OUTPUT_DIR,
    "external_test_results.json"
)


with open(
    external_results_path,
    "w"
) as f:

    json.dump(
        external_results,
        f,
        indent=4
    )


print(
    "\nHasil external test disimpan di:"
)

print(
    external_results_path
)

print(
    "\nConfusion matrix disimpan di:"
)

print(
    external_cm_path
)


print(
    "\n" + "=" * 70
)

print(
    "EXTERNAL TEST SELESAI."
)

print(
    "=" * 70
)
````

# Import Pustaka & Pembuatan Logika Grad-CAM

````python
import numpy as np
import tensorflow as tf
import matplotlib.pyplot as plt
import cv2
import os
import random

gradcam_preprocess_map = {
    'ResNet50': resnet_preprocess,
    'EfficientNetB0': efficientnet_preprocess,
    'MobileNetV2': mobilenet_preprocess,
    'DenseNet121': densenet_preprocess
}

gradcam_preprocess = gradcam_preprocess_map[best_model_name]

print("Grad-CAM siap digunakan.")


def find_last_conv_layer(model):

    for layer in reversed(model.layers):

        if isinstance(layer, tf.keras.layers.Conv2D) or layer.__class__.__name__ == 'Conv2D':
            return layer.name

        if isinstance(layer, tf.keras.Model) or hasattr(layer, 'layers'):

            for sublayer in reversed(layer.layers):

                if isinstance(sublayer, tf.keras.layers.Conv2D) or sublayer.__class__.__name__ == 'Conv2D':
                    return sublayer.name

    for layer in reversed(model.layers):
        if 'conv' in layer.name.lower():
            return layer.name

    raise ValueError(
        "Layer Conv2D tidak ditemukan."
    )


def make_gradcam_heatmap(
    img_array,
    model,
    last_conv_layer_name=None
):

    if last_conv_layer_name is None:

        last_conv_layer_name = find_last_conv_layer(
            model
        )

    last_conv_layer = model.get_layer(
        last_conv_layer_name
    )

    grad_model = tf.keras.models.Model(
        inputs=model.input,
        outputs=[
            last_conv_layer.output,
            model.output
        ]
    )

    with tf.GradientTape() as tape:

        conv_outputs, predictions = grad_model(
            img_array,
            training=False
        )

        pred_index = tf.argmax(
            predictions[0]
        )

        class_channel = predictions[
            :,
            pred_index
        ]

    grads = tape.gradient(
        class_channel,
        conv_outputs
    )

    pooled_grads = tf.reduce_mean(
        grads,
        axis=(0, 1, 2)
    )

    conv_outputs = conv_outputs[0]

    heatmap = conv_outputs @ tf.expand_dims(
        pooled_grads,
        axis=-1
    )

    heatmap = tf.squeeze(heatmap)

    heatmap = tf.maximum(
        heatmap,
        0
    )

    max_value = tf.reduce_max(heatmap)

    if float(max_value) != 0:

        heatmap /= max_value

    return heatmap.numpy()
````

# Fungsi Visualisasi Termal (Heatmap)

````python
print("2. Membangun fungsi visualisasi warna...")

def display_gradcam(img_path, heatmap, alpha=0.5):
    img = cv2.imread(img_path)
    img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
    img = cv2.resize(img, (224, 224))
    
    heatmap = cv2.resize(heatmap, (img.shape[1], img.shape[0]))
    heatmap = np.uint8(255 * heatmap)
    heatmap = cv2.applyColorMap(heatmap, cv2.COLORMAP_JET)
    
    superimposed_img = heatmap * alpha + img
    superimposed_img = np.clip(superimposed_img, 0, 255).astype('uint8')
    return img, heatmap, superimposed_img
````

````python
import matplotlib.pyplot as plt
import numpy as np
from sklearn.metrics import roc_curve, auc
from sklearn.preprocessing import label_binarize
from itertools import cycle

print("Menghasilkan Kurva Multiclass ROC-AUC...")

# Memastikan kita memiliki probabilitas mentah (bukan sekadar label tebakan)
final_test_generator.reset()

Y_pred_probs = final_model.predict(
    final_test_generator,
    verbose=1
)

Y_pred_probs = Y_pred_probs[
    :len(final_test_generator.classes)
]

# Binarize label asli (one-hot encoding format)
y_true_bin = label_binarize(
    final_test_generator.classes,
    classes=range(len(class_names))
)
n_classes = len(class_names)

# Hitung ROC curve dan ROC area untuk setiap kelas
fpr = dict()
tpr = dict()
roc_auc = dict()
for i in range(n_classes):
    fpr[i], tpr[i], _ = roc_curve(y_true_bin[:, i], Y_pred_probs[:, i])
    roc_auc[i] = auc(fpr[i], tpr[i])

# Plotting semua Kurva ROC
plt.figure(figsize=(10, 8))
colors = cycle(['aqua', 'darkorange', 'cornflowerblue', 'green', 'red', 'purple'])

for i, color in zip(range(n_classes), colors):
    plt.plot(fpr[i], tpr[i], color=color, lw=2,
             label='ROC curve of class {0} (area = {1:0.3f})'
             ''.format(class_names[i], roc_auc[i]))

# Garis diagonal (tebakan acak / worst case)
plt.plot([0, 1], [0, 1], 'k--', lw=2)

plt.xlim([0.0, 1.0])
plt.ylim([0.0, 1.05])
plt.xlabel('False Positive Rate', fontsize=12)
plt.ylabel('True Positive Rate', fontsize=12)
plt.title('Receiver Operating Characteristic (ROC) per Kelas', fontsize=14, fontweight='bold')
plt.legend(loc="lower right")
plt.grid(True, linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
````

````python
import os, shutil, json, gc
import numpy as np
import tensorflow as tf
from sklearn.model_selection import StratifiedKFold
from sklearn.utils.class_weight import compute_class_weight
from tensorflow.keras.optimizers import Adam
from tensorflow.keras.callbacks import EarlyStopping, ReduceLROnPlateau
from tensorflow.keras import regularizers, backend as K
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from scipy import stats

import hashlib

def get_base_name(fname):
    """
    Ekstrak group_id (nama dasar foto sebelum augmentasi) untuk mencegah data leakage.
    Roboflow menghasilkan varian augmentasi dengan format: <nama_asli>.rf.<hash>.<ekstensi>
    (Disamakan persis dengan fungsi di Cell 4 agar pengelompokan gambar konsisten)
    """
    if '.rf.' in fname:
        return fname.split('.rf.')[0]
    return os.path.splitext(fname)[0]

# Validasi konsistensi fungsi get_base_name
assert get_base_name("IMG_E4793_jpg.rf.66b3a30ca40300d98b0f71cd15f6068b.jpg") == "IMG_E4793_jpg", "Validasi get_base_name gagal!"
assert get_base_name("sample_photo.png") == "sample_photo", "Validasi fallback get_base_name gagal!"
print("✅ Fungsi get_base_name (group_id) konsisten dengan Cell 4.")

def compute_dataset_fingerprint(train_dir):
    """
    Menghasilkan fingerprint SHA-256 deterministik dari dataset training
    berdasarkan nama file dan ukuran file untuk mendeteksi jika isi dataset berubah.
    """
    hasher = hashlib.sha256()
    file_records = []
    for root, _, files in sorted(os.walk(train_dir)):
        for f in sorted(files):
            if f.lower().endswith(('.jpg', '.jpeg', '.png', '.bmp')):
                rel_path = os.path.relpath(os.path.join(root, f), train_dir)
                size = os.path.getsize(os.path.join(root, f))
                file_records.append(f"{rel_path}:{size}")
    hasher.update("\n".join(file_records).encode('utf-8'))
    return hasher.hexdigest()[:16]

# Pakai model terpilih + resep yang SAMA dengan eksperimen utama (Cell 8 dan 9)
cv_model_name = best_model_name             # hasil Cell 9
cv_config     = MODEL_CONFIGS[cv_model_name]  # dari Cell 8
cv_preprocess = cv_config['preprocess']

# ============================================================
# CROSS-VALIDATION MENGGUNAKAN DATASET FINAL (HANYA TRAINING SET)
# ============================================================

orig_train_dir = '/kaggle/working/dataset_split_clean/train'
assert os.path.exists(orig_train_dir), f"Direktori training tidak ditemukan: {orig_train_dir}"
assert "val" not in orig_train_dir and "test" not in orig_train_dir, "Cross-validation HANYA boleh menggunakan data training!"


all_groups = []
group_class = []
group_files = {}

# Kumpulkan group dari DATASET FINAL
for cls in sorted(os.listdir(orig_train_dir)):

    cls_path = os.path.join(orig_train_dir, cls)

    if not os.path.isdir(cls_path):
        continue

    for fname in os.listdir(cls_path):

        base = get_base_name(fname)

        if base not in group_files:
            group_files[base] = []

        group_files[base].append(
            os.path.join(cls_path, fname)
        )

# Pastikan setiap group hanya memiliki satu kelas
for base, files in group_files.items():

    classes = set(
        os.path.basename(os.path.dirname(f))
        for f in files
    )

    if len(classes) != 1:
        raise ValueError(
            f"Group {base} ditemukan pada lebih dari satu kelas: {classes}"
        )

    cls = list(classes)[0]

    all_groups.append(base)
    group_class.append(cls)

all_groups = np.array(all_groups)
group_class = np.array(group_class)

print("==============================================")
print("CROSS-VALIDATION DATASET")
print("==============================================")
print(f"Total group unik: {len(all_groups)}")

print("\nDistribusi group per kelas:")
for cls in sorted(np.unique(group_class)):
    print(
        f"{cls}: "
        f"{np.sum(group_class == cls)} group"
    )

all_groups = np.array(all_groups)
group_class = np.array(group_class)

# Validasi ketat bahwa tidak ada grup dari test set yang masuk ke cross-validation
test_dir_check = '/kaggle/working/dataset_split_clean/test'
if os.path.exists(test_dir_check):
    test_groups_set = {
        get_base_name(f) for _, _, files in os.walk(test_dir_check)
        for f in files if f.lower().endswith(('.jpg', '.jpeg', '.png', '.bmp'))
    }
    leakage_with_test = set(all_groups).intersection(test_groups_set)
    assert len(leakage_with_test) == 0, f"❌ LEAKAGE: {len(leakage_with_test)} grup test masuk ke cross-validation!"
print("✅ Validasi CV: Murni menggunakan data training (100% terisolasi dari data test).")


N_FOLDS = 5
CV_WORK_DIR = '/kaggle/working/cv_fold_data'
RESULTS_FILE = f'/kaggle/working/cv_fold_results_{cv_model_name}.json'

# Hitung fingerprint dataset training
dataset_fingerprint = compute_dataset_fingerprint(orig_train_dir)
print(f"Dataset fingerprint (SHA-256) : {dataset_fingerprint}")

# ============================================================
# VALIDASI & RESUME HASIL FOLD CROSS-VALIDATION
# ============================================================
FORCE_RERUN_CV = False  # Ubah ke True jika ingin memaksa ulang seluruh fold dari awal

cv_config_current = {
    "model_name": cv_model_name,
    "n_folds": N_FOLDS,
    "total_groups": int(len(all_groups)),
    "total_images": int(sum(len(group_files[b]) for b in all_groups)),
    "dataset_fingerprint": dataset_fingerprint
}

completed_results = []
if os.path.exists(RESULTS_FILE) and not FORCE_RERUN_CV:
    try:
        with open(RESULTS_FILE) as f:
            loaded_data = json.load(f)

        is_valid_resume = False
        # 1. Validasi format dengan metadata config & fingerprint dataset
        if isinstance(loaded_data, dict) and "config" in loaded_data:
            old_cfg = loaded_data["config"]
            model_match = old_cfg.get("model_name") == cv_model_name
            folds_match = old_cfg.get("n_folds") == N_FOLDS
            groups_match = old_cfg.get("total_groups") == len(all_groups)
            fp_match = old_cfg.get("dataset_fingerprint") == dataset_fingerprint

            if model_match and folds_match and groups_match and fp_match:
                completed_results = loaded_data.get("folds", [])
                is_valid_resume = True
                print(f"✅ Konfigurasi & fingerprint dataset cocok. Melanjutkan resume ({len(completed_results)}/{N_FOLDS} fold selesai).")
            else:
                print("⚠️ Terdeteksi perubahan konfigurasi atau dataset berubah!")
                if not fp_match:
                    print(f"   Fingerprint dataset berubah : lama={old_cfg.get('dataset_fingerprint')} vs baru={dataset_fingerprint}")
                if not model_match:
                    print(f"   Model arsitektur berubah    : lama={old_cfg.get('model_name')} vs baru={cv_model_name}")
                if not folds_match:
                    print(f"   Jumlah folds berubah        : lama={old_cfg.get('n_folds')} vs baru={N_FOLDS}")
                if not groups_match:
                    print(f"   Jumlah group berubah        : lama={old_cfg.get('total_groups')} vs baru={len(all_groups)}")
                print("🔄 Hasil fold lama direset agar tidak tercampur dengan eksperimen baru.")

        # 2. Validasi format list lama (legacy)
        elif isinstance(loaded_data, list):
            print("⚠️ File resume lama tidak memiliki fingerprint dataset. Reset hasil fold untuk menjamin integritas data.")

        if not is_valid_resume:
            completed_results = []
            if os.path.exists(RESULTS_FILE):
                os.remove(RESULTS_FILE)
    except Exception as e:
        print(f"⚠️ Gagal memvalidasi file resume ({e}). Memulai cross-validation baru.")
        completed_results = []
elif FORCE_RERUN_CV:
    print("🔄 FORCE_RERUN_CV aktif: Menghapus hasil lama dan menjalankan CV dari awal.")
    if os.path.exists(RESULTS_FILE):
        os.remove(RESULTS_FILE)

completed_folds = {r["fold"] for r in completed_results}

# StratifiedKFold dijalankan pada daftar GROUP (1 elemen = 1 group) => group-aware
skf = StratifiedKFold(n_splits=N_FOLDS, shuffle=True, random_state=42)

for fold_idx, (train_group_idx, val_group_idx) in enumerate(skf.split(all_groups, group_class)):
    fold_num = fold_idx + 1

    # --- Lewati fold yang sudah selesai di run sebelumnya ---
    if fold_num in completed_folds:
        print(f"Fold {fold_num}/{N_FOLDS} sudah selesai sebelumnya, dilewati.")
        continue

    print(f"\n{'='*50}\nFOLD {fold_num}/{N_FOLDS}\n{'='*50}")

    tr_groups = set(all_groups[train_group_idx])
    va_groups = set(all_groups[val_group_idx])
    assert tr_groups.isdisjoint(va_groups), "Group leakage pada fold!"
    n_tr_img = sum(len(group_files[b]) for b in tr_groups)
    n_va_img = sum(len(group_files[b]) for b in va_groups)
    print(f"Fold {fold_num}: train {len(tr_groups)} grup/{n_tr_img} gambar | val {len(va_groups)} grup/{n_va_img} gambar")

    fold_dir = os.path.join(CV_WORK_DIR, f'fold_{fold_idx}')
    if os.path.exists(fold_dir):
        shutil.rmtree(fold_dir)

    for idx_list, split_name in [(train_group_idx, 'train'), (val_group_idx, 'val')]:
        out_dir = os.path.join(fold_dir, split_name)
        for i in idx_list:
            base, cls = all_groups[i], group_class[i]
            dst_cls_dir = os.path.join(out_dir, cls)
            os.makedirs(dst_cls_dir, exist_ok=True)
            for fpath in group_files[base]:
                shutil.copy(fpath, os.path.join(dst_cls_dir, os.path.basename(fpath)))

    train_datagen_fold = ImageDataGenerator(
        preprocessing_function=cv_preprocess,
        rotation_range=20, zoom_range=0.2, horizontal_flip=True, fill_mode='nearest'
    )
    val_datagen_fold = ImageDataGenerator(preprocessing_function=cv_preprocess)

    train_gen_fold = train_datagen_fold.flow_from_directory(
        os.path.join(fold_dir, 'train'), target_size=(224, 224), batch_size=32,
        class_mode='categorical', shuffle=True, seed=42
    )
    val_gen_fold = val_datagen_fold.flow_from_directory(
        os.path.join(fold_dir, 'val'), target_size=(224, 224), batch_size=32,
        class_mode='categorical', shuffle=False
    )

    cls_w = compute_class_weight('balanced', classes=np.unique(train_gen_fold.classes), y=train_gen_fold.classes)
    class_weight_fold = dict(enumerate(cls_w))

    model_fold, base_fold = build_comparison_model(cv_config['base_class'])   # dari Cell 8

    # --- FASE 1: latih head saja, base model masih beku ---
    model_fold.fit(
        train_gen_fold, epochs=PHASE1_EPOCHS, validation_data=val_gen_fold,
        class_weight=class_weight_fold,
        callbacks=[EarlyStopping(monitor='val_loss', patience=3, restore_best_weights=True, verbose=0)],
        verbose=1
    )

    # --- FASE 2: fine-tuning, buka UNFREEZE_N layer terakhir ---
    base_fold.trainable = True
    for layer in base_fold.layers[:-UNFREEZE_N]:
        layer.trainable = False
    for layer in base_fold.layers[-UNFREEZE_N:]:
        layer.trainable = not isinstance(layer, tf.keras.layers.BatchNormalization)

    model_fold.compile(optimizer=Adam(learning_rate=LEARNING_RATE_PHASE2),
                       loss='categorical_crossentropy', metrics=['accuracy'])
    model_fold.fit(
        train_gen_fold, epochs=PHASE2_EPOCHS, validation_data=val_gen_fold,
        class_weight=class_weight_fold,
        callbacks=[EarlyStopping(monitor='val_loss', patience=4, restore_best_weights=True, verbose=0),
                   ReduceLROnPlateau(monitor='val_loss', factor=0.5, patience=2, min_lr=1e-7, verbose=0)],
        verbose=1
    )

    val_loss, val_acc = model_fold.evaluate(val_gen_fold, verbose=0)
    print(f"Fold {fold_num} validation accuracy: {val_acc*100:.2f}%")

    # --- Simpan hasil fold ini ke disk segera (beserta config eksperimen) ---
    completed_results.append({
        "fold": fold_num, "val_acc": float(val_acc), "val_loss": float(val_loss),
        "train_groups": len(tr_groups), "val_groups": len(va_groups),
        "train_images": n_tr_img, "val_images": n_va_img
    })
    save_payload = {
        "config": cv_config_current,
        "folds": completed_results
    }
    with open(RESULTS_FILE, 'w') as f:
        json.dump(save_payload, f, indent=2)
    os.makedirs('/kaggle/working/cinnamon-output', exist_ok=True)
    with open(f'/kaggle/working/cinnamon-output/cv_fold_results_{cv_model_name}.json', 'w') as f:
        json.dump(save_payload, f, indent=2)

    # --- Bersihkan data sementara & memori GPU sebelum fold berikutnya ---
    shutil.rmtree(fold_dir)
    del model_fold, base_fold, train_gen_fold, val_gen_fold
    K.clear_session()
    gc.collect()

# --- Ringkasan akhir dihitung dari file, bukan dari memory ---
with open(RESULTS_FILE) as f:
    final_cv_data = json.load(f)

if isinstance(final_cv_data, dict):
    completed_results = final_cv_data.get("folds", [])
else:
    completed_results = final_cv_data

fold_accuracies = np.array([r["val_acc"] for r in completed_results])

if len(fold_accuracies) < N_FOLDS:
    print(f"\n⚠️ Baru {len(fold_accuracies)}/{N_FOLDS} fold selesai. Jalankan cell ini lagi untuk melanjutkan fold yang tersisa.")
else:
    mean_acc = fold_accuracies.mean()
    std_acc = fold_accuracies.std(ddof=1)
    ci95 = stats.t.interval(0.95, len(fold_accuracies)-1, loc=mean_acc, scale=std_acc/np.sqrt(len(fold_accuracies)))

    print(f"\n{'='*50}\nHASIL {N_FOLDS}-FOLD CROSS VALIDATION\n{'='*50}")
    print(f"Akurasi per fold: {[f'{a*100:.2f}%' for a in fold_accuracies]}")
    print(f"Rata-rata: {mean_acc*100:.2f}% ± {std_acc*100:.2f}% (std)")
    print(f"95% CI: [{ci95[0]*100:.2f}%, {ci95[1]*100:.2f}%]")
````

````python
# ============================================================
# FINAL VISUALIZATION
# Internal Test + Perbandingan 4 Model
# ============================================================

import os
import json
import numpy as np
import tensorflow as tf
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.metrics import (
    confusion_matrix,
    classification_report,
    accuracy_score
)

from tensorflow.keras.preprocessing.image import (
    ImageDataGenerator
)


# ============================================================
# PATH
# ============================================================

RESULTS_FILE = (
    "/kaggle/working/model_comparison_results.json"
)

OUTPUT_DIR = (
    "/kaggle/working/visualisasi_hasil"
)

os.makedirs(
    OUTPUT_DIR,
    exist_ok=True
)


# ============================================================
# LOAD HASIL COMPARISON
# ============================================================

if not os.path.exists(
    RESULTS_FILE
):

    raise FileNotFoundError(
        f"File tidak ditemukan:\n"
        f"{RESULTS_FILE}"
    )


with open(
    RESULTS_FILE,
    "r"
) as f:

    comparison_results = json.load(f)


# ============================================================
# PILIH MODEL TERBAIK
# ============================================================

best_model_name = max(
    comparison_results,
    key=lambda x:
    comparison_results[x][
        "validation_accuracy"
    ]
)

checkpoint_path = (
    comparison_results[
        best_model_name
    ]["checkpoint"]
)


print("=" * 70)
print("FINAL VISUALIZATION")
print("=" * 70)

print(
    f"Model terbaik : "
    f"{best_model_name}"
)

print(
    f"Checkpoint    : "
    f"{checkpoint_path}"
)


# ============================================================
# LOAD MODEL
# ============================================================

best_model = tf.keras.models.load_model(
    checkpoint_path
)


# ============================================================
# INTERNAL TEST
# ============================================================

TEST_DIR = (
    "/kaggle/working/"
    "dataset_split_clean/test"
)

IMG_SIZE = (
    224,
    224
)

BATCH_SIZE = 32


# ============================================================
# PREPROCESSING INTERNAL
# ============================================================

if best_model_name == "ResNet50":

    from tensorflow.keras.applications.resnet50 import (
        preprocess_input
    )

elif best_model_name == "MobileNetV2":

    from tensorflow.keras.applications.mobilenet_v2 import (
        preprocess_input
    )

elif best_model_name == "DenseNet121":

    from tensorflow.keras.applications.densenet import (
        preprocess_input
    )

elif best_model_name == "EfficientNetB0":

    from tensorflow.keras.applications.efficientnet import (
        preprocess_input
    )

else:

    raise ValueError(
        f"Model tidak dikenali: "
        f"{best_model_name}"
    )


test_datagen = ImageDataGenerator(
    preprocessing_function=preprocess_input
)


test_gen = test_datagen.flow_from_directory(
    TEST_DIR,
    target_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    class_mode="categorical",
    shuffle=False
)


internal_class_names = list(
    test_gen.class_indices.keys()
)


print(
    "\nKelas Internal Test:"
)

print(
    internal_class_names
)


# ============================================================
# INTERNAL PREDICTION
# ============================================================

test_gen.reset()

y_prob_internal = best_model.predict(
    test_gen,
    verbose=1
)

y_pred_internal = np.argmax(
    y_prob_internal,
    axis=1
)

y_true_internal = (
    test_gen.classes
)


# ============================================================
# INTERNAL ACCURACY
# ============================================================

internal_accuracy = accuracy_score(
    y_true_internal,
    y_pred_internal
)


print(
    "\nInternal Test Accuracy:"
)

print(
    f"{internal_accuracy * 100:.2f}%"
)


# ============================================================
# INTERNAL CONFUSION MATRIX
# ============================================================

cm_internal = confusion_matrix(
    y_true_internal,
    y_pred_internal
)


plt.figure(
    figsize=(10, 8)
)

sns.heatmap(
    cm_internal,
    annot=True,
    fmt="d",
    cmap="Blues",
    cbar=True,
    linewidths=1,
    linecolor="white",
    xticklabels=internal_class_names,
    yticklabels=internal_class_names,
    annot_kws={
        "fontsize": 12,
        "fontweight": "bold"
    }
)

plt.title(
    f"Confusion Matrix - Internal Test\n"
    f"{best_model_name}",
    fontsize=14,
    fontweight="bold",
    pad=15
)

plt.xlabel(
    "Predicted Class",
    fontsize=12
)

plt.ylabel(
    "True Class",
    fontsize=12
)

plt.xticks(
    rotation=35,
    ha="right",
    fontsize=10
)

plt.yticks(
    rotation=0,
    fontsize=10
)

plt.tight_layout()


internal_cm_path = os.path.join(
    OUTPUT_DIR,
    "confusion_matrix_internal_test.png"
)


plt.savefig(
    internal_cm_path,
    dpi=300,
    bbox_inches="tight",
    facecolor="white"
)

plt.show()


# ============================================================
# INTERNAL CLASSIFICATION REPORT
# ============================================================

print(
    "\nClassification Report - Internal Test:"
)

print(
    classification_report(
        y_true_internal,
        y_pred_internal,
        target_names=internal_class_names,
        digits=4,
        zero_division=0
    )
)


# ============================================================
# PERBANDINGAN 4 MODEL
# ============================================================

model_names = []
validation_accuracies = []


for model_name, result in (
    comparison_results.items()
):

    model_names.append(
        model_name
    )

    validation_accuracies.append(
        result[
            "validation_accuracy"
        ] * 100
    )


# ============================================================
# SORT DARI TERTINGGI
# ============================================================

sorted_data = sorted(
    zip(
        model_names,
        validation_accuracies
    ),
    key=lambda x: x[1],
    reverse=True
)


model_names = [
    item[0]
    for item in sorted_data
]

validation_accuracies = [
    item[1]
    for item in sorted_data
]


# ============================================================
# BAR CHART
# ============================================================

plt.figure(
    figsize=(10, 6)
)

bars = plt.bar(
    model_names,
    validation_accuracies
)


plt.title(
    "Comparison of Validation Accuracy Across Four Architectures",
    fontsize=14,
    fontweight="bold",
    pad=15
)

plt.xlabel(
    "Model",
    fontsize=12
)

plt.ylabel(
    "Validation Accuracy (%)",
    fontsize=12
)

plt.ylim(
    0,
    100
)

plt.xticks(
    fontsize=10
)

plt.yticks(
    fontsize=10
)


# ============================================================
# NILAI DI ATAS BAR
# ============================================================

for bar, value in zip(
    bars,
    validation_accuracies
):

    plt.text(
        bar.get_x()
        + bar.get_width() / 2,

        value + 1,

        f"{value:.2f}%",

        ha="center",
        va="bottom",

        fontsize=11,
        fontweight="bold"
    )


plt.tight_layout()


comparison_path = os.path.join(
    OUTPUT_DIR,
    "perbandingan_4_model.png"
)


plt.savefig(
    comparison_path,
    dpi=300,
    bbox_inches="tight",
    facecolor="white"
)

plt.show()


# ============================================================
# RINGKASAN FINAL
# ============================================================

print(
    "\n" + "=" * 70
)

print(
    "RINGKASAN HASIL AKHIR"
)

print(
    "=" * 70
)


print(
    f"Best Model               : "
    f"{best_model_name}"
)


print(
    f"Validation Accuracy      : "
    f"{comparison_results[best_model_name]['validation_accuracy'] * 100:.2f}%"
)


print(
    f"Internal Test Accuracy   : "
    f"{internal_accuracy * 100:.2f}%"
)


# ============================================================
# EXTERNAL RESULT DARI CELL 10
# ============================================================

external_results_path = os.path.join(
    OUTPUT_DIR,
    "external_test_results.json"
)


if os.path.exists(
    external_results_path
):

    with open(
        external_results_path,
        "r"
    ) as f:

        external_results = json.load(f)


    print(
        f"External Test Accuracy  : "
        f"{external_results['external_accuracy'] * 100:.2f}%"
    )


    print(
        f"External Test Images    : "
        f"{external_results['number_of_images']}"
    )


print(
    "\nPerbandingan 4 model:"
)


for name, acc in sorted_data:

    print(
        f"{name:<20} : "
        f"{acc:.2f}%"
    )


# ============================================================
# OUTPUT FILE
# ============================================================

print(
    "\nFile visualisasi:"
)

print(
    f"- {internal_cm_path}"
)

print(
    f"- {comparison_path}"
)

print(
    f"- {external_results_path}"
)


print(
    "\n" + "=" * 70
)

print(
    "SEMUA VISUALISASI SELESAI."
)

print(
    "=" * 70
)

# ============================================================
# SIMPAN HASIL EVALUASI INTERNAL LENGKAP (6 KELAS)
# Dilakukan sebelum perbandingan fair agar tidak tertimpa hasil 4 kelas
# ============================================================
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

os.makedirs(OUTPUT_DIR, exist_ok=True)

internal_results_full = {
    "model": best_model_name,
    "classes": internal_class_names,
    "number_of_images": int(len(y_true_internal)),
    "accuracy": float(accuracy_score(y_true_internal, y_pred_internal)),
    "classification_report": classification_report(
        y_true_internal,
        y_pred_internal,
        target_names=internal_class_names,
        output_dict=True,
        zero_division=0
    ),
    "confusion_matrix": confusion_matrix(
        y_true_internal,
        y_pred_internal
    ).tolist()
}

with open(os.path.join(OUTPUT_DIR, "internal_test_results.json"), "w") as f:
    json.dump(internal_results_full, f, indent=4)

print(f"✅ internal_test_results.json (6 kelas) disimpan di {OUTPUT_DIR}")

# ============================================================
# PERBANDINGAN FAIR: INTERNAL (4 KELAS) VS EKSTERNAL (4 KELAS)
# Evaluasi 4 kelas SELALU dihitung ulang dari y_true_internal & y_pred_internal
# Tidak membaca file JSON lama agar angka selalu mencerminkan prediksi terbaru
# ============================================================
# Evaluasi internal untuk empat kelas yang sama dengan eksternal
cinnamon_4_classes = ["healthy", "leafspot", "roughbark", "stripecanker"]

# Normalisasi nama kelas untuk menghindari masalah huruf besar/kecil
class_name_map = {
    name.lower(): idx
    for idx, name in enumerate(internal_class_names)
}

missing = [
    name for name in cinnamon_4_classes
    if name not in class_name_map
]
if missing:
    raise ValueError(
        f"Kelas internal tidak ditemukan: {missing}. "
        f"Kelas tersedia: {internal_class_names}"
    )

# Indeks kelas asli (global, dari 6-kelas) dan pemetaan ke indeks lokal 0–3
global_indices = [
    class_name_map[name] for name in cinnamon_4_classes
]
global_to_local = {
    global_idx: local_idx
    for local_idx, global_idx in enumerate(global_indices)
}

# Ambil hanya gambar yang label aslinya termasuk empat kelas tersebut
mask_4c = np.isin(y_true_internal, global_indices)

# Petakan ke indeks lokal 0–3 agar y_true dan y_pred berada pada ruang yang sama
y_true_4c = np.array([
    global_to_local[lbl] for lbl in y_true_internal[mask_4c]
])
y_pred_4c = np.array([
    global_to_local.get(lbl, -1) for lbl in y_pred_internal[mask_4c]
])

# Gambar yang diprediksi sebagai kelas di luar 4 kelas target (misalnya noncinnamon_*) diberi label -1;
# filter keluar agar tidak mencemari metrik
valid_mask = y_pred_4c >= 0
y_true_4c = y_true_4c[valid_mask]
y_pred_4c = y_pred_4c[valid_mask]
n_pred_outside = int(np.sum(~valid_mask))
if n_pred_outside > 0:
    print(f"ℹ️ {n_pred_outside} gambar diprediksi di luar 4 kelas target — dikecualikan dari metrik 4-kelas.")

int_acc_4c_fresh = float(accuracy_score(y_true_4c, y_pred_4c)) if len(y_true_4c) > 0 else 0.0
print(f"\nAkurasi Internal (4 Kelas, dihitung ulang): {int_acc_4c_fresh * 100:.2f}%")
print(f"Sampel Internal yang dievaluasi (4 kelas) : {len(y_true_4c)} gambar")

# Simpan hasil 4 kelas sebagai file tersendiri, menimpa file lama jika ada
internal_4class_results = {
    "model": best_model_name,
    "classes": cinnamon_4_classes,
    "number_of_images_4classes": int(len(y_true_4c)),
    "internal_accuracy_4classes": float(int_acc_4c_fresh),
    "predictions_outside_4classes": n_pred_outside,
    "classification_report_4classes": classification_report(
        y_true_4c,
        y_pred_4c,
        labels=list(range(len(cinnamon_4_classes))),
        target_names=cinnamon_4_classes,
        output_dict=True,
        zero_division=0
    ),
    "confusion_matrix_4classes": confusion_matrix(
        y_true_4c,
        y_pred_4c,
        labels=list(range(len(cinnamon_4_classes)))
    ).tolist()
}

with open(os.path.join(OUTPUT_DIR, "internal_4class_results.json"), "w") as f:
    json.dump(internal_4class_results, f, indent=4)

print(f"✅ internal_4class_results.json (4 kelas) disimpan di {OUTPUT_DIR}")

# Periksa ketersediaan hasil evaluasi eksternal
ext_results_file = os.path.join(OUTPUT_DIR, "external_test_results.json")
ext_data = None

if os.path.exists(ext_results_file):
    try:
        with open(ext_results_file) as f:
            ext_data = json.load(f)
        # Validasi bahwa file eksternal adalah hasil eksperimen yang valid (ada kunci wajib)
        if 'external_accuracy' not in ext_data or 'model' not in ext_data:
            print("⚠️ File external_test_results.json tidak lengkap. Evaluasi eksternal perlu dijalankan ulang.")
            ext_data = None
        elif ext_data.get('model') != best_model_name:
            print(f"⚠️ File evaluasi eksternal berasal dari model '{ext_data.get('model')}', bukan '{best_model_name}'. Jalankan Cell 10 ulang.")
            ext_data = None
    except Exception as e_ext:
        print(f"⚠️ Gagal membaca external_test_results.json: {e_ext}")
        ext_data = None
else:
    print("⚠️ File external_test_results.json belum tersedia. Pastikan Cell 10 telah dieksekusi sebelum Cell ini.")

# Lakukan perbandingan fair hanya jika evaluasi eksternal valid & dari eksperimen yang sama
if ext_data is not None:
    int_acc_4c = int_acc_4c_fresh * 100
    ext_acc_4c = ext_data.get('external_accuracy', 0.0) * 100

    print("\n" + "=" * 70)
    print("PERBANDINGAN AKURASI: INTERNAL (4 KELAS) VS EKSTERNAL (4 KELAS)")
    print("=" * 70)
    print(f"Akurasi Internal (4 Kelas Sama)  : {int_acc_4c:.2f}%")
    print(f"Akurasi Eksternal (4 Kelas Sama) : {ext_acc_4c:.2f}%")
    print(f"Selisih Generalisasi (Gap)       : {abs(int_acc_4c - ext_acc_4c):.2f}%")

    # Bar chart perbandingan internal vs eksternal
    plt.figure(figsize=(8, 6))
    categories = ['Internal Test\n(4 Kelas Sama)', 'External Test\n(4 Kelas Sama)']
    acc_values = [int_acc_4c, ext_acc_4c]
    bars = plt.bar(categories, acc_values, color=['#2b5c8f', '#e67e22'], width=0.5)
    plt.title(
        f"Fair Comparison: Internal vs External Test\n"
        f"({best_model_name} - 4 Cinnamon Disease Classes)",
        fontsize=13, fontweight='bold', pad=15
    )
    plt.ylabel("Accuracy (%)", fontsize=12)
    plt.ylim(0, 100)

    for bar in bars:
        h = bar.get_height()
        plt.text(bar.get_x() + bar.get_width() / 2.0, h + 1.5, f"{h:.2f}%",
                 ha='center', va='bottom', fontsize=12, fontweight='bold')

    plt.tight_layout()
    comp_plot_path = os.path.join(OUTPUT_DIR, 'perbandingan_internal_external_4kelas.png')
    plt.savefig(comp_plot_path, dpi=300, bbox_inches='tight')
    plt.show()
    print(f"✅ Grafik perbandingan 4 kelas disimpan di {comp_plot_path}")

    # Simpan ringkasan fair comparison (menimpa file lama)
    fair_summary = {
        'model': best_model_name,
        'classes': cinnamon_4_classes,
        'internal_accuracy_4classes': float(int_acc_4c),
        'external_accuracy_4classes': float(ext_acc_4c),
        'generalization_gap': float(abs(int_acc_4c - ext_acc_4c))
    }
    with open(os.path.join(OUTPUT_DIR, 'internal_vs_external_4classes.json'), 'w') as f:
        json.dump(fair_summary, f, indent=4)
    with open('/kaggle/working/internal_vs_external_4classes.json', 'w') as f:
        json.dump(fair_summary, f, indent=4)
    print("✅ Ringkasan perbandingan fair 4 kelas disimpan di internal_vs_external_4classes.json")
else:
    print("\n" + "=" * 70)
    print("ℹ️ Perbandingan fair internal–eksternal dilewati: evaluasi eksternal belum tersedia.")
    print("   Jalankan Cell 10 terlebih dahulu, kemudian jalankan kembali cell ini.")
    print("=" * 70)
````

````python
# ============================================================
# EKSPOR SELURUH HASIL EKSPERIMEN KE ZIP
# Mengumpulkan model, checkpoints model_backups, metrik JSON, grafik, dan prediksi
# Dilengkapi pembuatan manifest_hasil_eksperimen.json & uji integritas ZIP
# ============================================================

import os
import json
import zipfile
import datetime

OUTPUT_ZIP = '/kaggle/working/hasil_eksperimen.zip'
MANIFEST_FILE = '/kaggle/working/manifest_hasil_eksperimen.json'
print("Mengumpulkan seluruh hasil eksperimen ke dalam ZIP...")

# Hapus ZIP lama jika sudah ada
if os.path.exists(OUTPUT_ZIP):
    os.remove(OUTPUT_ZIP)

collected_items = []
seen_rel_paths = set()

# Folder target yang dikumpulkan
target_folders = [
    'visualisasi_hasil',
    'model_comparison_plots',
    'model_backups',
    'cinnamon-output'
]

# Folder dataset gambar mentah yang HARUS DIKECUALIKAN agar ZIP tidak membengkak
EXCLUDED_DIR_NAMES = {
    'final_cinnamon-1',
    'dataset_split_clean',
    'cinnamon-multiple-diseases-1',
    'cinnamon-multiple-diseases',
    'cv_fold_data',
    '.git',
    '__pycache__'
}

# Ekstensi umum hasil evaluasi & visualisasi
STANDARD_RESULT_EXTENSIONS = (
    '.json', '.csv', '.log', '.png', '.jpg', '.jpeg', '.svg', '.txt'
)

# Ekstensi checkpoint model (termasuk yang non-standar)
MODEL_CHECKPOINT_EXTENSIONS = (
    '.keras', '.h5', '.weights.h5', '.ckpt', '.weights', '.pb', '.pt', '.pth', '.bin', '.onnx', '.tflite'
)

def is_valid_artifact_file(folder_name, root, fname):
    """
    Menentukan apakah suatu file merupakan artefak eksperimen yang harus disertakan.
    Memastikan file checkpoint model yang tidak berekstensi umum tetap terarsip.
    """
    f_lower = fname.lower()
    if fname.endswith('.zip') or fname.startswith('.'):
        return False
        
    # 1. Seluruh file di dalam folder model_backups SELALU disertakan
    #    (termasuk file checkpoint tanpa ekstensi, metadata json, dll)
    if 'model_backups' in root.split(os.sep):
        return True

    # 2. File checkpoint model di manapun (berdasarkan ekstensi atau penamaan checkpoint/ckpt)
    if f_lower.endswith(MODEL_CHECKPOINT_EXTENSIONS) or f_lower.startswith(('checkpoint', 'ckpt-')):
        return True

    # 3. File hasil visualisasi, grafik, metrik JSON, dan CSV
    if f_lower.endswith(STANDARD_RESULT_EXTENSIONS):
        return True

    return False

# ============================================================
# 1. PENGUMPULAN FILE ARTEFAK KE DALAM ZIP
# ============================================================
with zipfile.ZipFile(OUTPUT_ZIP, 'w', zipfile.ZIP_DEFLATED) as zipf:
    # A. Kumpulkan file dari direktori target
    for folder_name in target_folders:
        folder_path = os.path.join('/kaggle/working', folder_name)
        if os.path.exists(folder_path):
            for root, dirs, files in os.walk(folder_path):
                # Filter agar folder dataset mentah tidak ikut dijelajahi
                dirs[:] = [d for d in dirs if d.lower() not in EXCLUDED_DIR_NAMES and not d.startswith('.')]
                
                parts = [p.lower() for p in root.split(os.sep)]
                if any(ex in parts for ex in EXCLUDED_DIR_NAMES):
                    continue

                for f in files:
                    if not is_valid_artifact_file(folder_name, root, f):
                        continue
                        
                    full_p = os.path.join(root, f)
                    rel_p = os.path.relpath(full_p, '/kaggle/working')
                    if rel_p not in seen_rel_paths:
                        zipf.write(full_p, rel_p)
                        seen_rel_paths.add(rel_p)
                        collected_items.append((rel_p, os.path.getsize(full_p)))

    # B. Kumpulkan file penting di root /kaggle/working
    for fname in sorted(os.listdir('/kaggle/working')):
        fpath = os.path.join('/kaggle/working', fname)
        if os.path.isfile(fpath) and is_valid_artifact_file('root', '/kaggle/working', fname):
            if fname not in seen_rel_paths:
                zipf.write(fpath, fname)
                seen_rel_paths.add(fname)
                collected_items.append((fname, os.path.getsize(fpath)))

# ============================================================
# 2. PEMERIKSAAN KELENGKAPAN HASIL EVALUASI & MODEL
# ============================================================
print("\n" + "=" * 70)
print("PEMERIKSAAN KELENGKAPAN ARTEFAK EKSPERIMEN:")
print("=" * 70)

expected_jsons = [
    ('model_comparison_results.json', 'Hasil perbandingan 4 arsitektur'),
    ('internal_test_results.json', 'Hasil evaluasi internal test'),
    ('external_test_results.json', 'Hasil evaluasi eksternal test'),
    ('internal_vs_external_4classes.json', 'Hasil perbandingan fair 4 kelas')
]

missing_evaluations = []
for j_file, desc in expected_jsons:
    in_zip = any(j_file in item[0] for item in collected_items)
    if in_zip:
        print(f"  [✅ TERSIMPAN] {j_file:<35} : {desc}")
    else:
        print(f"  [❌ HILANG   ] {j_file:<35} : {desc}")
        missing_evaluations.append(j_file)

# Periksa file cross-validation JSON
cv_jsons = [item[0] for item in collected_items if 'cv_fold_results' in item[0]]
if cv_jsons:
    for cv_f in cv_jsons:
        print(f"  [✅ TERSIMPAN] {cv_f:<35} : Hasil Cross-Validation")
else:
    print("  [ℹ️ INFO    ] cv_fold_results_*.json belum ditemukan (opsional, jika CV dijalankan).")

# Periksa model & checkpoint (termasuk yang di model_backups)
model_files = [
    item for item in collected_items 
    if item[0].lower().endswith(MODEL_CHECKPOINT_EXTENSIONS) or 'model_backups' in item[0] or item[0].lower().startswith(('checkpoint', 'ckpt-'))
]

print(f"\nModel & Checkpoint tersimpan: {len(model_files)} file")
for m_path, m_size in model_files:
    print(f"  - {m_path} ({m_size / (1024 * 1024):.2f} MB)")

# Peringatan jika model tidak ditemukan
if len(model_files) == 0:
    print("\n⚠️ PERINGATAN KERAS: Tidak ada file model atau checkpoint yang ditemukan di dalam arsip!")

# Peringatan jika evaluasi penting belum lengkap
if missing_evaluations:
    print("\n" + "!" * 70)
    print("⚠️ PERINGATAN: HASIL EVALUASI BELUM LENGKAP!")
    print("!" * 70)
    print("File JSON penting berikut belum tersedia di dalam arsip:")
    for mf in missing_evaluations:
        print(f"  ❌ {mf}")
    print("💡 Disarankan untuk menjalankan cell evaluasi yang terlewat sebelum mengunduh ZIP.")
    print("!" * 70)
else:
    print("\n✅ Seluruh file evaluasi utama lengkap dan terverifikasi.")

# ============================================================
# 3. PEMBUATAN MANIFEST ARTEFAK HASIL EKSPERIMEN
# (zip_integrity tidak disertakan di sini — akan diuji pada step akhir setelah manifest masuk ZIP)
# ============================================================
zip_size_mb = os.path.getsize(OUTPUT_ZIP) / (1024 * 1024)

manifest_data = {
    "timestamp": datetime.datetime.now().isoformat(),
    "zip_filename": "hasil_eksperimen.zip",
    "zip_size_mb": round(zip_size_mb, 2),
    "total_files": len(collected_items),
    "evaluation_status": "INCOMPLETE" if missing_evaluations else "COMPLETE",
    "missing_evaluations": missing_evaluations,
    "models_count": len(model_files),
    "models": [m[0] for m in model_files],
    "files": [
        {"path": path, "size_bytes": sz, "size_mb": round(sz / (1024 * 1024), 3)}
        for path, sz in collected_items
    ]
}

with open(MANIFEST_FILE, 'w') as mf:
    json.dump(manifest_data, mf, indent=2)

# Tambahkan manifest ke dalam ZIP
with zipfile.ZipFile(OUTPUT_ZIP, 'a', zipfile.ZIP_DEFLATED) as zipf:
    zipf.write(MANIFEST_FILE, 'manifest_hasil_eksperimen.json')

print(f"✅ Manifest disimpan ke {MANIFEST_FILE} dan disisipkan ke dalam ZIP.")

# ============================================================
# 4. UJI INTEGRITAS ZIP FINAL
# Dilakukan SETELAH manifest masuk sehingga menguji arsip yang benar-benar final
# ============================================================
print("\n" + "=" * 70)
print("PENGUJIAN INTEGRITAS ZIP FINAL:")
print("=" * 70)

with zipfile.ZipFile(OUTPUT_ZIP, "r") as final_zip:
    corrupted_member = final_zip.testzip()

if corrupted_member is not None:
    raise RuntimeError(
        f"❌ ZIP rusak! File bermasalah terdeteksi: {corrupted_member}\n"
        "Harap jalankan kembali cell ini untuk membuat ulang arsip."
    )

final_zip_size_mb = os.path.getsize(OUTPUT_ZIP) / (1024 * 1024)
print("✅ ZIP final valid: Seluruh file (termasuk manifest) lolos uji CRC32.")
print(f"📦 Lokasi  : {OUTPUT_ZIP}")
print(f"💾 Ukuran  : {final_zip_size_mb:.2f} MB")
print(f"📄 Total   : {len(collected_items) + 1} file (termasuk manifest)")
print("=" * 70)
print("🎉 Arsip siap diunduh dari panel Files Kaggle.")
print("=" * 70)
````
